```python
# Write a program to count the number of vowels and consonants present in an input string.
def count_vow_cons(input_str):
    vow = set("aeiouAEIOU")
    vow_count = 0
    cons_count = 0
    for char in input_str:
        if char.isalpha():
            if char in vow :
                vow_count += 1
            else :
                cons_count += 1
    return vow_count , cons_count
input_str = "Machine learning"
result = count_vow_cons(input_str)
print("Number of vowels = " , result[0])
print("Number of consonents = " , result[1])
```

```python
# Write a program that accepts two matrices A and B as input and returns their product AB. Check if A & B are multipliable; if not, return error message. 
def mul_mat(mat_a , mat_b):
    rows_a = len(mat_a)
    cols_a = len(mat_a[0])
    rows_b = len(mat_b)
    cols_b = len(mat_b[0])
    if cols_a != rows_b:
        return "Error : Matrices can't be multiplied coz cols of A and rows of B are equal"
    result_mat = [[0 for _ in range(cols_b)] for _ in range(rows_a)]
    for i in range(rows_a):
        for j in range(cols_b):
            for k in range(cols_a):
                result_mat[i][j] += mat_a[i][k] * mat_b[k][j]
    return result_mat
mat_1 = [[1,2,3] , [4,5,6]]
mat_2 = [[7,8],[9,10],[11,12]]
result = mul_mat(mat_1 , mat_2)
if isinstance(result , str):
    print(result)
else :
    print("The product matrix AB = ")
    for row in result :
        print(row)
```

```python
# Write a program to find the number of common elements between two lists. The lists 
contain integers. 
def count_same_num(list_a , list_b):
    set_a = set(list_a)
    set_b = set(list_b)
    same_num = set_a.intersection(set_b)
    total_count = len(same_num)
    return total_count
list_1 = [1,2,3,4,5]
list_2 = [4,5,6,7,8]
result = count_same_num(list_1 , list_2)
print("The common elements = " , result)
```

```python
# Write a program that accepts a matrix as input and returns its transpose.
def transpose_mat(mat):
    rows = len(mat)
    cols = len(mat[0])
    transposed = [[0 for _ in range(rows)] for _ in range(cols)]
    for i in range(rows):
        for j in range(cols):
            transposed[j][i] = mat[i][j]
    return transposed
mat = [[1,2,3] , [4,5,6]]
result = transpose_mat(mat)
if isinstance(result , str):
    print(result)
else :
    print("The transposed matrix = ")
    for row in result :
        print(row)
```

```python
import random
from collections import Counter
def calculation(num_list) :
    mean_val = sum(num_list) / len(num_list)
    sort_num = sorted(num_list)
    total_num = len(sort_num)
    mid_idx_1 = (total_num // 2) - 1
    mid_idx_2 = total_num // 2
    median_val = (sort_num[mid_idx_1] + sort_num[mid_idx_2]) / 2
    freq_counter = Counter(num_list)
    max_freq = max(freq_counter.values())
    mode_val = [num for num , count in freq_counter.items() if count == max_freq]
    return mean_val , median_val , mode_val
random_num = [random.randint(100, 150) for _ in range(100)]
mean, median, mode = calculation(random_num)
print("Generate 100 Random Numbers:")
print(random_num)
print("-" * 50)
print("Mean :", mean)
print("Median :", median)
print("Mode :", mode)
```
